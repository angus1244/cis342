```python
from itertools import combinations as iter_combinations

text = "project"
n = len(text)

all_combinations = [
  "".join(text[i] for i in idx_tuple)
  for r in range(1, n + 1)
  for idx_tuple in iter_combinations(range(n), r)
]

print(all_combinations)
```
